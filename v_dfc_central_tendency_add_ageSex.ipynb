```python
df_final = pd.merge(
    merged_df, metadata_subset, left_on="ID", right_on="Observations", how="left"
)
```

```python
cols = list(df_final.columns)
new_column_order = cols[0:1] + cols[-2:] + cols[1:-2]
df_final = df_final[new_column_order]
```

```python
# delete the "Observations" column
df_final = df_final.drop(columns=["Observations"])
```

```python
# rename "gender_text" column to "sex"
df_final.rename(columns={"gender_text": "sex"}, inplace=True)
# sort dataframe by "age" 
df_final.sort_values(by="age", inplace=True)
```
